```python
## 영어단장 프로그램 입니다.
## 영어단어 입력 후 한국어 뜻을 입력하면 vocabulary.txt 파일을 저장합니다.
## q를 입력하면 프로그램이 종료 됩니다.

def vocabulary_input():
    word_list = []
    i = 0

    while True:
        en_word = input("영어 단어를 입력하세요: ").strip()
        if en_word.lower() == 'q':
            break
        kr_word = input("단어 뜻을 입력하세요: ").strip()
        if kr_word.lower() == 'q':
            break
        word_list.append((en_word, kr_word))
        i += 1
        print(f"{i}번 단어 저장됨: ({en_word, kr_word})\n")
    
    return word_list

def save_to_file(word_list, filename="vocabulary.txt"):
    #header = ["English Word", "Korean Meaning"]
    with open(filename, "w", encoding="utf-8") as f:
        for en_word, kr_word in word_list:
            f.write(f"{en_word}: {kr_word} \n")
    print("저장 완료:", word_list)

if __name__ == "__main__":
    word_list = vocabulary_input()
    save_to_file(word_list)
```

```python
# vocabulary.txt을 읽어, 영어 단어를 보여주고, 뜻을 입력 받아
# 맞다면, 맞았습니다. 
# 틀렸다면, 아쉽습니다. 정답은 xxx입니다. 라고 출력한다.
# ocabulary.txt 입력된 순서로 진행하고 저장된 내용이 끝이면 끝
def read_voca_list(filename):
    word_list = []
    try:
        with open( filename, mode='r', encoding='utf-8') as f:
            for line in f:
                line = line.strip()
                if line and ":" in line:  
                #f에서 읽어온 line변수에 값이 있고(&), ":" 값이 있다면
                    word, meaning = line.split(":",1) #':'기준으로 나누고 1번만 실시
                    word_list.append((word,meaning))
            return word_list 
    except FileNotFoundError:
        print(f"오류: '{filename}'파일을 찾을 수 없습니다.")
        return None

def quiz_voca(word_list):
    for word, meaning in word_list:
        user_meaning = input(f"{word} 뜻을 입력하세요: ").strip()
        print(f'입력하신 값은: {user_meaning}')
        if user_meaning.lower() == 'q': break
        if user_meaning == meaning.strip():
            print(f"{meaning} 맞았습니다!")
        else:
            print(f"아쉽습니다. 정답은 '{meaning}'입니다. ")
    print("테스트 종료합니다.")
    return None

if __name__ == "__main__":
    voca_list = read_voca_list('vocabulary.txt')
    quiz_voca(voca_list)


```
