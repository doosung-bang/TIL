```python
##영어 단어 및 한국어 뜻 입력받아 vocabulary.txt 저장하는 프로그램
## 2026.08.16

## 1.입력받는 def 정의
## 2.입력받은 정보 저장하는 def
```

```python
def input_voca():
    word_list=[]
    i = 0
    while True:
        i += 1
        en_word = str(input('영어 단어를 입력하세요: ')).strip()
        if en_word.lower() == 'q': 
            print('입력을 종료 하셨습니다.') 
            break
        kr_meaning = input('한국어 뜻을 입력하세요: ').strip()
        if kr_meaning.lower() == 'q': 
            print('입력을 종료 하셨습니다.') 
            break

        word_list.append((en_word, kr_meaning))
        print(f'{i}번째 {en_word}: {kr_meaning} 단어가 입력되었습니다.')
    return word_list

def write_voca(word_list, filename='data/vocabulary.txt'):
    with open(filename, mode='a', encoding='utf-8') as f:
        for e, k in word_list:
            f.write(f'{e}: {k}\n')
    print(f'단장이 저장되었습니다. 파일:{filename}')

if __name__ == "__main__":
    word_list = input_voca()
    write_voca(word_list)
```

```python
#########요청 사항
## vocabulary.txt 값을 읽어오고 
## 사용자에게 영어 단어를 보여주고 한글 뜻을 입력 받아 비교하여
## 맞으면 "정답입니다!" 
## 틀리면 "아쉽니다." "정답은 xxx 입니다."
## 특이 사항: 사용자에게 보여줄 때 파일의 숫서가 아닌 랜덤하게 보여주기

#구현 사항
#1. def 정의: 파일 읽어 오기
#2. def 정의: 사용자에게 랜덤하게 보여주고 정답확인


def read_voca(filename='data/vocabulary.txt'):
    word_dict={}
    try:
        with open(filename, mode='r', encoding='utf-8') as f:
            for line in f:
                if line and ":" in line:
                    data = line.split(":",1) #line을 split하고 윈쪽 2개의 변수에 저장
                    word_dict[data[0].strip()] = data[1].strip() 
                    print(data[0], ":", data[1] ,"\n")
                    #사전의 저장 방식: "사전이름[키] = 데이터"
        return word_dict
    except FileNotFoundError:
        print(f"오류: '{filename}'파일을 찾을 수 없습니다.")
        return False
    
import random
def start_quiz(vocab_dict):

    if not vocab_dict:
        print("퀴즈를 시작할 단어가 없습니다.")
        return False

    words = list(vocab_dict.keys())
    random.shuffle(words)

    score = 0
    total = len(words)

    print("==== 퀴즈 시작 ====")
    for word in words:
        correct_answer = vocab_dict[word]
        user_answer = input(f"{word}의 뜻은?").strip()
        if user_answer.lower() == 'q': break

        if user_answer == correct_answer:
            print(f"{correct_answer} 정답입니다.")
            score += 1
        else:
            print(f"아쉽니다. 정답은 '{correct_answer}'입니다.")
    print(f"\n최종 점수는: {score}/{total}")
    return True

if __name__ == "__main__":
    voca_list = read_voca()
    start_quiz(voca_list)


                        
```
