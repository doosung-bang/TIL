```python
#대출 신청자들을 여러 개의 그룹으로 나누려고 합니다. 
# 일단 대출이 승낙된 사람들 중에서 자영업자가 아니고, 수입이 중간값 이상인 사람들을 추출하려고 하는데요. 
# 불린 인덱싱을 사용해서 세 가지 조건을 모두 만족하는 데이터를 group1이라는 변수에 저장해 주세요.

import pandas as pd

loan_df = pd.read_csv('data/loan.csv')
loan_df = loan_df.set_index('loan_id')
new_columns = {'married_or_not': 'married',
               'self_employed_or_not': 'self_employed',
               'applicant_income': 'income',
               'loan_amount': 'amount'                
}
##컬럼 이름 변경
loan_df = loan_df.rename(columns=new_columns)
####income컬럼 형변환
loan_df['income'].astype('float')
loan_df
```

```python
mask1 = (loan_df['loan_status'] == 'Y') & (loan_df['self_employed'] == 'N')
mask2 = loan_df['income'] >= loan_df['income'].median()
group1 = loan_df[mask1 & mask2]
group1.info()
loan_df['income'].describe()
```
