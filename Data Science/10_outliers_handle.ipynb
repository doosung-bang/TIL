```python
##효준이는 스마트폰 구매 후보에서 가격이 너무 싸거나 비싼 건 제외하려고 합니다. 
# cellphone_df에서 가격을 기준으로 이상점으로 볼 수 있는 데이터는 삭제해 주세요.

import pandas as pd
cellphone_df = pd.read_csv('data/cellphone.csv')
cellphone_df
```

```python
q1 = cellphone_df['price'].quantile(0.25)
q3 = cellphone_df['price'].quantile(0.75)
iqr = q3 - q1
lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr
cellphone_df[(cellphone_df['price'] < lower_limit) | (cellphone_df['price'] > upper_limit)]
```

```python
mask = cellphone_df[(cellphone_df['price'] < lower_limit) | (cellphone_df['price'] > upper_limit)].index
```

```python
cellphone_df.drop(mask)
```

```python
cellphone_df = cellphone_df.drop(mask)
cellphone_df
```
