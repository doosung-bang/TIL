```python
import pandas as pd

airbnb_df = pd.read_csv('data/airbnb.csv')

airbnb_df.info()
airbnb_df.duplicated().sum()
```

```python
print(airbnb_df.duplicated().sum())
airbnb_df[airbnb_df.duplicated()] #boolean값으로 인덱싱
```

```python
#전체 중복값 삭제
airbnb_df.drop_duplicates()
airbnb_df = airbnb_df.drop_duplicates()

print(airbnb_df.duplicated().sum())
```

```python
airbnb_df[airbnb_df.duplicated(subset='id')] 
```

```python
airbnb_df[airbnb_df.duplicated(subset=['id','price' ]) ]
```

```python
airbnb_df[airbnb_df.duplicated(subset='id', keep=False)]  ##keep = 'First'  첫번째는 유지 , keep=False 전체
```
