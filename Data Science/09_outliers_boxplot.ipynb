```python
import pandas as pd

airbnb_df = pd.read_csv('data/airbnb.csv')

airbnb_df.info()
airbnb_df

```

```python
airbnb_df.describe()
```

```python
q1 = airbnb_df['price'].quantile(0.25)
q3 = airbnb_df['price'].quantile(0.75)
iqr = q3 - q1
lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr
```

```python
airbnb_df[airbnb_df['price'] < lower_limit]  #--> 가격이 너무 낮은 이상점 정보
```

```python
airbnb_df[airbnb_df['price'] > upper_limit]  #--> 가격이 너무 높은 이상점 정보
```

```python
airbnb_df[ (airbnb_df['price'] < lower_limit) | (airbnb_df['price'] > upper_limit) ]
```

```python
## 이상점 처리 방법
# 1. 이상점 정보 삭제
# 2. 이상점 정보를 최저 또는 최고 limit 으로 변경
# 3. 이상점 정보를 제외하고, 정상치 정보만 취급
# 
#### 이상점 제외하고 정상 정보만 처리 방법 
condition1 = airbnb_df['price'] >= lower_limit
condition2 = airbnb_df['price'] <= upper_limit
airbnb_df[condition1 & condition2]
```

```python
## 이상점을 제외한 정보만 다시 저장
airbnb_df = airbnb_df[condition1 & condition2] 
airbnb_df
```
