```python
import pandas as pd

loan_df = pd.read_csv('data/loan.csv')
loan_df = loan_df.set_index('loan_id')
new_columns = {'married_or_not': 'married',
               'self_employed_or_not': 'self_employed',
               'applicant_income': 'income',
               'loan_amount': 'amount'                
}
##컬럼 이름 변경
loan_df = loan_df.rename(columns=new_columns)
####income컬럼 형변환
loan_df['income'].astype('float')
loan_df
```

```python
income_mean = loan_df['income'].mean()
df = loan_df.query("(married == 'Y') & (income > @income_mean) & (property_area not in ['City', 'Suburb'])") 
df
```

```python
df = df.drop('LP2945')  # axis=0 x행에서 인덱스 LP1006,  axis='index'같은 의미
df = df.drop('married', axis=1) # 컬럼: married 삭제 ,  axis = 'columns' 같은 의미
df = df.drop(columns='self_employed') # 컬럼: self_employed 삭제
df 
```
