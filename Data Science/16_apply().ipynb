## pandas의 cut() 함수를 사용해서 사람들의 연령대를 분류해 봤습니다. 
이번 시간에는 apply() 라는 함수에 대해 알아보고, 이 함수를 가지고 똑같은 작업을 다시 해 볼 건데요. 

cut()이 데이터 구간화를 위해 만들어진 함수라면, 
apply()는 원하는 기능을 마음대로 구현할 수 있는 만능 함수라고 할 수 있어요.

일단 apply()는 다른 함수를 아규먼트로 사용합니다. apply가 '적용하다'라는 뜻을 가지고 있는 것처럼 DataFrame이나 Series에 있는 값 하나하나에 아규먼트로 받은 함수를 적용해 주는데요.

```python
import pandas as pd

df = pd.DataFrame([[1, 9], [4, 16]], columns=['column1', 'column2'])
df
```

여기서 column2에 있는 9와 16의 제곱근 값을 각각 구해 볼게요. 이때 제곱근을 계산해 주는 numpy의 sqrt()라는 함수를 사용할 건데요. 아래와 같이 sqrt() 안에 9라는 값을 넘겨주면 9의 제곱근인 3.0을 리턴합니다.

```python
import numpy as np

np.sqrt(9)  # 출력 내용 -> 3.0
```

이제 sqrt()를 apply()와 함께 사용해 볼게요. column2에 있는 값들의 제곱근을 구할 거니까, 아래와 같이 df['column2'].apply()라고 써 준 뒤 괄호 안에는 numpy의 sqrt() 함수를 넣어 주면 됩니다.

```python
df['column2'].apply(np.sqrt)
```

여기서 주의할 점은, 함수 안에 함수를 넘겨줄 때에는 함수의 이름까지만 쓰고 뒤에 괄호를 쓰지 않아야 한다는 건데요. 그러니까 np.sqrt라고 써야지 np.sqrt()라고 쓰면 안 된다는 거죠. 위 코드를 실행하면 아래 보이는 것처럼 3과 4, 즉 9와 16의 제곱근 값을 담고 있는 Series가 나옵니다.

## apply()로 데이터 구간화하기

```python
patient_df = pd.read_csv('data/patient.csv')
patient_df.head()
```

```python
def group_age(x):
    if x >= 10 and x < 20:
        return '10s'
    elif x >= 20 and x < 30:
        return '20s'
    elif x >= 30 and x < 40:
        return '30s'
    elif x >= 40 and x < 50:
        return '40s'
    elif x >= 50 and x < 60:
        return '50s'
    else:
        return '60s'

patient_df['age_group'] = patient_df['age'].apply(group_age)
patient_df
```

## lambda (람다) 함수와 apply()

마지막으로 한 가지만 더 알아 볼게요. apply()는 람다 함수라는 특이한 형태의 함수와 함께 사용하기도 하는데요. 람다 함수는 def라는 키워드를 사용해서 함수를 따로 정의할 필요 없이, 다른 함수 안에서 바로 사용할 수 있습니다. 따로 이름이 없다는 뜻으로 '익명 함수(Anonymous Function)'이라고 부르기도 하죠.

```python
def multiply_two(x):
    return 2*x
```

```python
lambda x: 2*x
```

이제 apply()와 람다 함수를 함께 사용해 볼게요. patient_df의 height 컬럼에는 미터 기준 키 정보가 들어 있었는데요. 표기 단위를 미터에서 인치로 바꾸고 싶다면, 1미터는 약 39.370079인치니까 height 값에 39.370079를 곱해 주면 되겠죠? 아래와 같은 코드를 통해, apply()로 Series에 있는 값들에 하나씩 접근해서 람다 함수를 적용할 수 있어요.

```python
patient_df['height'].apply(lambda x: 39.370079*x)
```

방금 전에는 람다 함수의 기본 사용법을 알아보기 위해 엄청 간단한 코드를 예로 들었는데요. 사실 이런 경우에는 굳이 람다 함수를 사용할 필요 없이 아래 코드처럼 직접 컬럼에 사칙연산을 적용하는 것이 좋습니다. 코드도 좀 더 간결하고 실행 속도도 더 빠르기 때문이죠.

```python
patient_df['height'] * 39.370079
```

그리고 if문을 함께 사용하면 람다 함수를 좀 더 유용하게 사용할 수 있습니다. 예를 들어 아래와 같이 bmi라는 컬럼에 BMI 값을 계산해 두고, BMI 값이 30이 넘으면 비만('Y'), 넘지 않으면 비만이 아니라고('N') 구분한다고 할게요. 그러려면 콜론 뒤에 바로 조건식이 참일 때 리턴할 값인 Y를 써 주고, if문을 쓴 다음에는 else 뒤에 조건식이 거짓일 때 리턴할 값인 N을 써 주면 됩니다.

```python
patient_df['bmi'] = round(patient_df['weight'] / patient_df['height']**2, 1)
print(patient_df['bmi'])
patient_df['bmi'].apply(lambda x: 'Y' if x > 30 else 'N')
```
