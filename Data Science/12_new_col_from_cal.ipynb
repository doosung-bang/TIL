## 비만도를 나타내는 지표로 'BMI(Body Mass Index)'라는 것도 많이 사용합니다. 
BMI를 계산하려면 몸무게(kg)를 키(m)의 제곱으로 나눠 주면 되는데요. 
예를 들어, 키가 180cm이고, 몸무게가 70kg인 사람의 BMI는 70을 1.8의 제곱으로 나눈 값인 21.6입니다.

위 내용을 참고해서 patient_df에 있는 각 환자의 BMI 값을 계산해서 bmi라는 컬럼에 저장해 주세요. 단, BMI 값은 소수 첫째 자리까지 반올림해 주셔야 합니다.

```python
import pandas as pd 

patient_df = pd.read_csv('data/patient.csv')

patient_df.dtypes
patient_df
```

```python
patient_df['bmi'] = round(patient_df['weight'] / patient_df['height']**2 , 1)
patient_df
```
