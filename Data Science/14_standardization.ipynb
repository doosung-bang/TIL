## weight 컬럼에 있는 값들이 평균은 0, 표준편차는 1이 되도록 표준화(Standardization)를 해 주세요.

```python
import pandas as pd 

patient_df = pd.read_csv('data/patient.csv')
```

```python
patient_df['weight'] = (patient_df['weight'] - patient_df['weight'].mean()) \
                        / patient_df['weight'].std()
```
