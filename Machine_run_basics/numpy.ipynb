-- 행렬의 덧샘과 곱샘 
A * B

NumPy에서 * 연산자는 요소별 곱셈을 수행해요. 
하지만 행렬곱은 @ 연산자를 사용하거나 np.dot() 함수를 사용해요. 따라서 모든 행렬 곱이 요소별 곱셈으로 동작하지는 않아요. 요소별 곱셈은 같은 위치의 요소들끼리 곱하는 연산이라, 두 행렬의 차원이 같아야 해요.

행렬 곱셈이 가능하려면 앞 행렬의 열 개수와 뒤 행렬의 행 개수가 같아야 합니다. 이때 결과 행렬의 크기는 앞 행렬의 행 개수 × 뒤 행렬의 열 개수가 됩니다

예> (m × n) 행렬 × (p × q) 행렬 
-->  n == p 숫자가 같으면 곱셈 가능
--> 곱의 차원 결과: m x q 


```python
import numpy as np

A = np.array([
    [1, -1, 2],
    [3, 2, 2,],
    [ 4, 1, 2]
])

B = np.random.rand(3,3)

display(A),display(B)

display(A * 5)
```

```python
# 행렬 4개 연산 연습
import numpy as np

A = np.array([
    [1, -1, 2],
    [3, 2, 2]
])

B = np.array([
    [0, 1],
    [-1, 1],
    [5, 2]
])

C = np.array([
    [2, -1],
    [-3, 3]
])

D = np.array([
    [-5, 1],
    [2, 0]
])

# 행렬 연산 결과를 result에 저장하세요
a = (2 * A) @ (-1 * B) 
result =  (2 * A) @ (-1 * B) @ (3*C + D)
result = 2*A @ -B @ (3*C + D)
# 테스트 코드
result

```

15. 전치 행렬, 단위 행렬, 역행렬

전치행렬 Transposed matrix:  행렬의 행과 열을 서로 바꾼 행렬을 의미합니다. 
- m×n 행렬의 전치행렬은 n×m 행렬이 되며, 기호로는 A^T로 나타냅니다. (대각선 기준 전환)
[[1, 2, 1],
 [3, 2, 2]
]
==>
[[1, 3]
 [2, 3]
 [1, 2]
]

단위행렬 Identity matrix : 주대각선 성분은 모두 1이고 나머지 성분은 모두 0인 정사각행렬 (정사각형 구조)
- 2 * 1 = 2 ==> 선형대수학에서 일반수학 1의 역활

혁행렬 Inverse matrix
- 10 * 1/10 = 1 , 5 * 1/5 = 1
- 곱했을 때 1이 나오도록 하는 수를 '역수'라고 하는데, 선형대수학에서는 역행렬이라고 한다. (정사각형 구조)


```python
#종합해서 복잡한 행렬 연산을 numpy를 사용해서 계산

import numpy as np

A = np.array([
    [1, -1, 2],
    [3, 2, 2]
])

B = np.array([
    [0, 1],
    [-1, 1],
    [5, 2]
])

C = np.array([
    [2, -1],
    [-3, 3]
])

D = np.array([
    [-5, 1],
    [2, 0]
])

# B(T) x (2A(T)) x (3C(-1) + D(T))
# 행렬 연산을 result 변수에 저장하세요
result = B.T @ (2*A.T) @ (3* np.linalg.pinv(C) + D.T)
#result = B.T @ (2*A.T) @ (3*np.linalg.pinv(C) + D.T)

# 테스트 코드
result
```
