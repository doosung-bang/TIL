```python
import pandas as pd 
import numpy as np 

from sklearn import preprocessing 
```

```python
filepath = 'data/NBA_player_of_the_week.csv'
nb_player = pd.read_csv(filepath)
display(nb_player.info())
display(type(nb_player))
display(nb_player.head())
nb_player.describe(include='all')
```

```python
height_weight_age_df = nb_player[['Height CM', 'Weight KG', 'Age']]
type(height_weight_age_df)
```

```python
from sklearn import preprocessing
import pandas as pd 
import numpy as np 

filepath = 'data/NBA_player_of_the_week.csv' 
nba_player = pd.read_csv(filepath)
#pd.set_option('display.float_format', lambda x : '%.5f' %x)

height_weight_age_df = nba_player[['Height CM', 'Weight KG', 'Age' ]]
type(height_weight_age_df)
##표준화
scaler = preprocessing.StandardScaler()
std_data = scaler.fit_transform(height_weight_age_df )
display(std_data)
std_df = pd.DataFrame(std_data, columns=['Height', 'Weight', 'Age' ])
display(std_df)
std_df.describe()
```
